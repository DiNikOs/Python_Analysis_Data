Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

```python
# Создаём функцию для преобразования в новый признак
def get_education(str):
    # Разбиваем строку по пробелам и берем первые два слова
    str = ' '.join(str.split(' ')[:2])
    
    if str == 'Высшее образование':
        return 'высшее'
    elif str == 'Неоконченное высшее':
        return 'неоконченное высшее'
    elif str == 'Среднее специальное':
        return 'среднее специальное'
    elif str == 'Среднее образование':
        return 'среднее'

# Применяем функцию и создаем новый столбец
df['Образование'] = df['Образование и ВУЗ'].apply(get_education)
# Проверяем, что получилось ровно 4 нужные категории
print(df['Образование'].unique())
print(df['Образование'].value_counts())
# Удаляем признак(столбец) 
df = df.drop('Образование и ВУЗ', axis=1)
df.head(3)
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
# Выполняем преобразование строки в количество месяцев работы
def get_experience(str):
    # Проверяем на NaN или 'Не указано'
    if pd.isna(str) or str == 'Не указано':
        return np.nan
    
    # Разбиваем строку по пробелам и берем срез слов (пропускаем слова "Опыт" и "работы", берем 2 пары "число + слово")
    words = str.split(' ')[2:6]    
    year_words = ['год', 'года', 'лет']
    month_words = ['месяц', 'месяца', 'месяцев']
    
    years = 0
    months = 0    
    for i in range(len(words)):
        if words[i] in year_words:
            years = int(words[i - 1])
        elif words[i] in month_words:
            months = int(words[i - 1])
            
    return years * 12 + months

# Создаем новый признак и удаляем старый
df['Опыт работы (месяц)'] = df['Опыт работы'].apply(get_experience)
print('Медианный опыт работы (мес.):', df['Опыт работы (месяц)'].median())

df = df.drop('Опыт работы', axis=1)
df.head(3)
```

Выполните преобразования и удалите столбец **"Город, переезд, командировки"** из таблицы.

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
# На всякий случай делаем проверку уникальных вариантов переезда
def get_relocation_clean(str):
    # Разбиваем всю строку по запятой на список частей       
    for part in str.split(','):
        # В части текста ищем кусок 'перее' и если находим - вытаскиваем всё что до '('
        if 'перее' in part:   
            return part.split('(')[0].strip()
            
    return None
# Выводим подсчет всех уникальных формулировок переезда
print(df['Город, переезд, командировки'].apply(get_relocation_clean).value_counts(dropna=False))

# Города милионники
million_cities = [
    'Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань', 
    'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 
    'Красноярск', 'Пермь', 'Воронеж', 'Волгоград'
]
# Готовность к перезду
relocation_ready = {
    'готов к переезду', 
    'готова к переезду', 
    'хочу переехать'
}
# Готовность к командировкам
trip_ready = {
    'готов к командировкам', 
    'готова к командировкам', 
    'готов к редким командировкам', 
    'готова к редким командировкам'
}
# Определяем города
def get_city(str):   
    city = str.split(',')[0].split('(')[0].strip()
    
    if city == 'Москва' or city == 'Санкт-Петербург':
        return city
    elif city in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'

# Общая функция по условиям (первый split по запятой, второй по '(')  
def get_readiness_by_conditions(str, condition, set_conditions):
    for part in str.split(','):
            if condition in part:
                status = part.split('(')[0].strip()
                return status in set_conditions
    return False  
# Функция для определения готовности к переезду
def get_relocation(str): 
    return get_readiness_by_conditions(str, 'перее', relocation_ready)
# Функция для определения готовности к командировкам
def get_business_trip(str):
    return get_readiness_by_conditions(str, 'командировк', trip_ready)

# Создаем три новых столбца
df['Город'] = df['Город, переезд, командировки'].apply(get_city)
df['Готовность к переезду'] = df['Город, переезд, командировки'].apply(get_relocation)
df['Готовность к командировкам'] = df['Город, переезд, командировки'].apply(get_business_trip)

# Ответы на контрольные вопросы
spb_percent = round(df['Город'].value_counts(normalize=True)['Санкт-Петербург'] * 100)
ready_percent = round(df[df['Готовность к переезду'] & df['Готовность к командировкам']].shape[0] / df.shape[0] * 100)
print('Проживают в Санкт-Петербурге (%):', spb_percent)
print('Готовы и к переезду, и к командировкам (%):', ready_percent)

df = df.drop('Город, переезд, командировки', axis=1)
df.head(3)
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).
На сайте hh.ru соискатель может указывать различные комбинации данных категорий, например:
* полная занятость, частичная занятость
* частичная занятость, проектная работа, волонтерство
* полный день, удаленная работа
* вахтовый метод, гибкий график, удаленная работа, полная занятость

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
# Тип занятости
employments = [
    'полная занятость', 
    'частичная занятость', 
    'проектная работа', 
    'волонтерство', 
    'стажировка'
]
# График работы
charts = [
    'полный день', 
    'сменный график', 
    'гибкий график', 
    'удаленная работа', 
    'вахтовый метод'
]

# Создаем признаки-мигалки для занятости
for employment in employments:
    df[employment] = df['Занятость'].apply(lambda x: employment in x)
# Создаем признаки-мигалки для графика работы
for chart in charts:
    df[chart] = df['График'].apply(lambda x: chart in x)
    
# Ответы на контрольные вопросы
proj_vol = df[df['проектная работа'] & df['волонтерство']].shape[0]
shift_flex = df[df['вахтовый метод'] & df['гибкий график']].shape[0]
print('Проектная работа и волонтерство:', proj_vol)
print('Вахтовый метод и гибкий график:', shift_flex)
    
df = df.drop(['Занятость', 'График'], axis=1)
df.head(3)
```

Например, для казахстанского тенге курс на 20.08.2019 составляет 17.197 руб. за 100 тенге, тогда итоговый курс равен - 17.197 / 100 = 0.17197 руб за 1 тенге.
Воспользуйтесь этой информацией в ваших преобразованиях.

Осталось только понять, откуда брать дату, по которой определяется курс? А вот же она - в признаке **"Обновление резюме"**, в нем содержится дата и время, когда соискатель выложил текущий вариант своего резюме. Нас интересует только дата, по ней бы и будем сопоставлять курсы валют.

Теперь у нас есть вся необходимая информация для того, чтобы создать признак "ЗП (руб)" - заработная плата в рублях.

После ответа на контрольные вопросы удалите исходный столбец заработной платы "ЗП" и все промежуточные столбцы, если вы их создавали.

Итак, давайте обсудим возможный алгоритм преобразования: 
1. Перевести признак "Обновление резюме" из таблицы с резюме в формат datetime и достать из него дату. В тот же формат привести признак "date" из таблицы с валютами.
2. Выделить из столбца "ЗП" сумму желаемой заработной платы и наименование валюты, в которой она исчисляется. Наименование валюты перевести в стандарт ISO согласно с таблицей выше.
3. Присоединить к таблице с резюме таблицу с курсами по столбцам с датой и названием валюты (подумайте, какой тип объединения надо выбрать, чтобы в таблице с резюме сохранились данные о заработной плате, изначально представленной в рублях). Значение close для рубля заполнить единицей 1 (курс рубля самого к себе)
4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
# Загружаем таблицу с курсами валют
rates = pd.read_csv('data/ExchangeRates.csv')
rates.head(3)
# Приводим дату в таблице курсов к стандартному виду без времени
rates['date'] = pd.to_datetime(rates['date'], format='%d/%m/%y').dt.date

# Словарь перевода валют в ISO-кодировку по таблице из задания
currency_map = {
    'руб': 'RUB',
    'грн': 'UAH',
    'белруб': 'BYN',
    'сум': 'UZS'
}

# Выделяем из столбца "ЗП" сумму и валюту
def get_salary_num(str):
    if pd.isna(str):
        return np.nan
    # Берем первое слово до пробела и переводим в число с плавающей точкой
    num_str = str.split(' ')[0]
    return float(num_str)

def get_salary_currency(str):
    if pd.isna(str):
        return np.nan
    # Берем второе слово (валюту) и убираем точку в конце, если она есть
    curr_str = str.split(' ')[1].replace('.', '')  
    
    # Если валюта есть в словаре замен — возвращаем ISO-код, иначе оставляем как есть (USD, EUR и др.)
    if curr_str in currency_map:
        return currency_map[curr_str]
    else:
        return curr_str

# Создаем временные столбцы с суммой, валютой и датой обновления
df['salary_sum'] = df['ЗП'].apply(get_salary_num)
df['currency'] = df['ЗП'].apply(get_salary_currency)
df['Обновление резюме'] = pd.to_datetime(df['Обновление резюме'], dayfirst=True).dt.date

# Объединяем основную таблицу с курсами по валюте и дате
df = df.merge(
    rates[['currency', 'date', 'proportion', 'close']], 
    left_on=['currency', 'Обновление резюме'],
    right_on=['currency', 'date'], 
    how='left'
)

# Для российского рубля (RUB) курс и пропорция отсутствуют в файле — заполняем единицами
df['close'] = df['close'].fillna(1)
df['proportion'] = df['proportion'].fillna(1)

# Вычисляем итоговую зарплату в рублях. Записывается всё в формате float64.
# Decimal в нашем случае не подходит, так как pandas и библиотеки графиков (matplotlib, seaborn, plotly) не умеют напрямую рисовать графики по типу Decimal.
df['ЗП (руб)'] = df['salary_sum'] * df['close'] / df['proportion']

# Ответ на контрольный вопрос
median_salary_rub = round(df['ЗП (руб)'].median() / 1000)
print('Медианная ЗП (тыс. руб.):', median_salary_rub)

# Удаляем исходный столбец "ЗП" и все промежуточные столбцы
cols_to_drop = ['ЗП', 'salary_sum', 'currency', 'date', 'proportion', 'close']
df = df.drop(cols_to_drop, axis=1)
df.head(3)
```

7. Постройте сводную таблицу, иллюстрирующую зависимость **медианной** желаемой заработной платы от возраста (**"Возраст"**) и образования (**"Образование"**). На полученной сводной таблице постройте **тепловую карту**. Проанализируйте тепловую карту, сравнив показатели внутри групп.

```python
# Строим сводную таблицу зависимости медианной ЗП от образования и возраста
pivot = df.pivot_table(
    values='ЗП (руб)',
    index='Образование',
    columns='Возраст',
    aggfunc='median'
)

# Выстраиваем уровни образования по логическому порядку (от высшего к среднему)
edu_order = ['высшее', 'неоконченное высшее', 'среднее специальное', 'среднее']
pivot = pivot.reindex(edu_order)

# Выводим ключевые расчеты по каждой группе образования
print("Анализ показателей по уровням образования:")
for edu in edu_order:
    row = pivot.loc[edu].dropna()
    max_sal = row.max()     # ищет самое большое число среди значений (находит самую высокую зарплату
    max_age = row.idxmax()  # ищет максимальное значение индекса, которым здесь стал возраст
    mean_sal = row.mean()
    print(f" - {edu}: пик {max_sal:,.0f} руб. (в {max_age} лет) | среднее по всем возрастам: {mean_sal:,.0f} руб.")
print("-" * 75)

# Строим тепловую карту
fig = plt.figure(figsize=(16, 5))
heat = sns.heatmap(
    data=pivot,
    cmap='YlGnBu',
    cbar_kws={'label': 'Медианная зарплата (руб.)'}
)

heat.set_title('Тепловая карта медианной заработной платы по образованию и возрасту', fontsize=16)
heat.set_xlabel('Возраст (лет)')
heat.set_ylabel('Уровень образования')
plt.show()
```

- Сравнение с группами без высшего образования: У соискателей со средним специальным и средним образованием зарплатный потолок наступает значительно раньше и находится существенно ниже.
```python
# Логарифмируем признак 'Возраст' (в данных нет нулевых значений возраста, поэтому +1 не требуется)
log_age = np.log(df['Возраст'])

# Вычисляем среднее (mu) и стандартное отклонение (sigma)
mu = log_age.mean()
sigma = log_age.std()

# Границы метода 3х и 4х сигм для визуализации на графике
lower_bound_3 = mu - 3 * sigma
upper_bound_3 = mu + 3 * sigma
upper_bound_4 = mu + 4 * sigma

# Строим гистограмму распределения в логарифмическом масштабе
fig, ax = plt.subplots(figsize=(10, 5))
histplot = sns.histplot(log_age, bins=30, ax=ax)

# Добавляем вертикальные линии среднего и границ трех сигм
histplot.axvline(mu, color='k', lw=2, label='Среднее (mu)')
histplot.axvline(lower_bound_3, color='k', ls='--', lw=2, label='Нижняя граница (-3 sigma)')
histplot.axvline(upper_bound_3, color='k', ls='--', lw=2, label='Верхняя граница (+3 sigma)')
histplot.axvline(upper_bound_4, color='r', ls=':', lw=2, label='Граница с послаблением (+4 sigma)')

histplot.set_title('Логарифмическое распределение возраста соискателей')
histplot.set_xlabel('Логарифм возраста (ln)')
histplot.set_ylabel('Количество резюме')
histplot.legend()

# Поиск и удаление выбросов методом z-отклонений (3 сигмы влево, 4 сигмы вправо)
def outliers_z_score_mod(data, feature, left=3, right=4, log_scale=True):
    if log_scale:
        x = np.log(data[feature] + 1)
    else:
        x = data[feature]            
    mu_val = x.mean()
    sigma_val = x.std()
    lower = mu_val - left * sigma_val
    upper = mu_val + right * sigma_val
    
    outliers = data[(x < lower) | (x > upper)]
    cleaned = data[(x >= lower) & (x <= upper)]
    return outliers, cleaned

emissions, df = outliers_z_score_mod(df, 'Возраст', left=3, right=4, log_scale=True)
plt.show()

print(f'Число выбросов по методу z-отклонения с границей послабления справа в 4 сигмы: {emissions.shape[0]}')
print(f'Результирующее число записей в таблице: {df.shape}')

# Выводим таблицу с найденными выбросами
display(emissions[['Возраст', 'ЗП (руб)', 'Образование', 'Опыт работы (месяц)', 'Ищет работу на должность:']])
```
